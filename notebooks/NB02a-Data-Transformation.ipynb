```python
import json
import pandas as pd
```

```python
with open("../data/raw/tmdb_major_movies.json", "r") as f:
    major_tmdb= json.load(f)
```

```python
major_df = pd.DataFrame(major_tmdb)
major_df.head()
```

```python
major_df = major_df.drop(["production_companies","id"], axis=1)
major_df.head()
```

```python
major_df["studio_type"]="major"
major_df.head()
```

Same for indie company

```python
with open("../data/raw/tmdb_indie_movies.json", "r") as f:
    indie_movies = json.load(f)

indie_df = pd.DataFrame(indie_movies)
indie_df = indie_df.drop(["production_companies","id"], axis=1)
indie_df["studio_type"] = "indie"
indie_df.head()
```

把indie和major拼起来

```python
tmdb_df = pd.concat(
    [major_df,indie_df],
    ignore_index = True
)
tmdb_df.head()
tmdb_df.tail()
```
